```python

import pandas as pd

# Read the student dataset
df = pd.read_csv('/content/StudentsPerformance.csv')

# Display first 5 rows
df.head()
```

```python

# Check the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns)

# Check dataset information
print("\nDataset Information:")
df.info()
```

```python

# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python

# Calculate average marks
print("Average Math Score:", df['math score'].mean())

print("Average Reading Score:", df['reading score'].mean())

print("Average Writing Score:", df['writing score'].mean())
```

```python

# Calculate total marks
df['total_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
)

# Calculate average marks
df['average_score'] = df['total_score'] / 3

# Display the results
print("Student Performance:")
df[['math score', 'reading score',
    'writing score', 'total_score',
    'average_score']].head()
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(df['average_score'], bins=10)

plt.title("Student Average Score Distribution")
plt.xlabel("Average Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Calculate average marks in each subject

subjects = ['math score', 'reading score', 'writing score']

subject_averages = df[subjects].mean()

print("Subject-wise Average Performance:")
print(subject_averages)
```

```python

import matplotlib.pyplot as plt

subjects = ['Math', 'Reading', 'Writing']
scores = [66.089, 69.169, 68.054]

plt.figure(figsize=(8,5))

plt.bar(subjects, scores)

plt.title("Subject-wise Average Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Score")

plt.show()
```

```python

# Gender-wise student performance

gender_performance = df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()

print("Gender-wise Performance:")
print(gender_performance)
```

```python

gender_performance.plot(kind='bar', figsize=(8,5))

plt.title("Gender-wise Student Performance")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.show()
```

```python

# Performance based on test preparation course

preparation_performance = df.groupby('test preparation course')[
    ['math score', 'reading score', 'writing score']
].mean()

print("Test Preparation Course Performance:")
print(preparation_performance)
```

```python
preparation_performance.plot(kind='bar', figsize=(8,5))

plt.title("Performance Based on Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.show()
```
